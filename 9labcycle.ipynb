```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.reshape(-1, 784) / 255.0
x_test = x_test.reshape(-1, 784) / 255.0

y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

```python
labeled_fraction = 0.1  # 10% labeled
num_labeled = int(len(x_train) * labeled_fraction)

x_labeled = x_train[:num_labeled]
y_labeled = y_train_cat[:num_labeled]

x_unlabeled = x_train
```

```python
input_img = Input(shape=(784,))
encoded1 = Dense(256, activation='relu')(input_img)
decoded1 = Dense(784, activation='sigmoid')(encoded1)

ae1 = Model(input_img, decoded1)
ae1.compile(optimizer='adam', loss='binary_crossentropy')

ae1.fit(x_unlabeled, x_unlabeled, epochs=10, batch_size=256)
```

```python
encoded_input = Input(shape=(256,))
encoded2 = Dense(128, activation='relu')(encoded_input)
decoded2 = Dense(256, activation='relu')(encoded2)

ae2 = Model(encoded_input, decoded2)
ae2.compile(optimizer='adam', loss='binary_crossentropy')

features1 = Model(input_img, encoded1).predict(x_unlabeled)
ae2.fit(features1, features1, epochs=10, batch_size=256)
```

```python
encoder = Model(
    input_img,
    Dense(128, activation='relu')(
        Dense(256, activation='relu')(input_img)
    )
)
```

```python
encoded_features = encoder(input_img)
output = Dense(10, activation='softmax')(encoded_features)

classifier = Model(input_img, output)
classifier.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
classifier.fit(
    x_labeled, y_labeled,
    epochs=20,
    batch_size=128,
    validation_data=(x_test, y_test_cat)
)
```

```python
loss, accuracy = classifier.evaluate(x_test, y_test_cat)
print("Test Accuracy:", accuracy)
```
