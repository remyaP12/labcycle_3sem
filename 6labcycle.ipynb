6.Design LSTM-based Recurrent Neural Networks to forecast time series data such as stock prices or weather trends.

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import MinMaxScaler
```

```python
# Dummy stock price data
data = np.array([
    100, 102, 101, 105, 108, 110, 112, 111, 115, 118,
    120, 122, 121, 125, 128, 130
]).reshape(-1, 1)
```

```python
scaler = MinMaxScaler(feature_range=(0, 1))
data_scaled = scaler.fit_transform(data)
```

```python
def create_sequences(data, time_steps=3):
    X, y = [], []
    for i in range(len(data) - time_steps):
        X.append(data[i:i+time_steps])
        y.append(data[i+time_steps])
    return np.array(X), np.array(y)

TIME_STEPS = 3
X, y = create_sequences(data_scaled, TIME_STEPS)
```

```python
model = Sequential([
    LSTM(50, activation='tanh', input_shape=(TIME_STEPS, 1)),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)

model.summary()
```

```python
model.fit(
    X, y,
    epochs=50,
    batch_size=1,
    verbose=1
)
```

```python
last_sequence = data_scaled[-TIME_STEPS:]
last_sequence = last_sequence.reshape(1, TIME_STEPS, 1)

predicted_price = model.predict(last_sequence)
predicted_price = scaler.inverse_transform(predicted_price)

print("Predicted next value:", predicted_price[0][0])
```
