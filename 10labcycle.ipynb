10.Implement a custom CNN with optimizer (SGD, RMSProp, Adam) inorder to analyze the influence of the same in training time and accuracy of a deep neural network.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.datasets import mnist
import time
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.reshape(-1, 28, 28, 1) / 255.0
x_test = x_test.reshape(-1, 28, 28, 1) / 255.0
```

```python
def create_cnn(optimizer):
    model = Sequential([
        Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
        MaxPooling2D((2,2)),
        Conv2D(64, (3,3), activation='relu'),
        MaxPooling2D((2,2)),
        Flatten(),
        Dense(128, activation='relu'),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
start = time.time()
model_sgd = create_cnn(tf.keras.optimizers.SGD(learning_rate=0.01))
model_sgd.fit(x_train, y_train, epochs=5, batch_size=128, verbose=1)
sgd_time = time.time() - start

sgd_acc = model_sgd.evaluate(x_test, y_test, verbose=0)[1]
```

```python
start = time.time()
model_rms = create_cnn(tf.keras.optimizers.RMSprop())
model_rms.fit(x_train, y_train, epochs=5, batch_size=128, verbose=1)
rms_time = time.time() - start

rms_acc = model_rms.evaluate(x_test, y_test, verbose=0)[1]
```

```python
start = time.time()
model_adam = create_cnn(tf.keras.optimizers.Adam())
model_adam.fit(x_train, y_train, epochs=5, batch_size=128, verbose=1)
adam_time = time.time() - start

adam_acc = model_adam.evaluate(x_test, y_test, verbose=0)[1]
```

```python
print("\nOptimizer Comparison:")
print(f"SGD     → Accuracy: {sgd_acc:.4f}, Time: {sgd_time:.2f}s")
print(f"RMSProp → Accuracy: {rms_acc:.4f}, Time: {rms_time:.2f}s")
print(f"Adam    → Accuracy: {adam_acc:.4f}, Time: {adam_time:.2f}s")
```
