7.Perform object classification or detection using CNN using TensorFlow or Keras with a pre-trained model like MobileNet or SSD on COCO or custom dataset.



```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNet
from tensorflow.keras.applications.mobilenet import preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
```

```python
model = MobileNet(weights='imagenet')
```

```python
img = image.load_img('/content/pexels-david-vives-1474251-28990268.jpg', target_size=(224, 224))
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)
img_array = preprocess_input(img_array)
```

```python
predictions = model.predict(img_array)
decoded = decode_predictions(predictions, top=3)

for label in decoded[0]:
    print(label)
```
