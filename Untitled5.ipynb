```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.layers import BatchNormalization
from tensorflow.keras.datasets import cifar10
```

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# Normalize pixel values (0–255 → 0–1)
x_train = x_train / 255.0
x_test = x_test / 255.0

print(x_train.shape, y_train.shape)
```

```python
model_no_norm = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)),
    MaxPooling2D(2,2),

    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),

    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_no_norm.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nTraining CNN without Normalization")
model_no_norm.fit(x_train, y_train, epochs=5, batch_size=64)

loss, acc_no = model_no_norm.evaluate(x_test, y_test)
print("Test Accuracy (No Normalization):", acc_no)
```

```python
model_bn = Sequential([
    Conv2D(32, (3,3), input_shape=(32,32,3)),
    BatchNormalization(),
    tf.keras.layers.Activation('relu'),
    MaxPooling2D(2,2),

    Conv2D(64, (3,3)),
    BatchNormalization(),
    tf.keras.layers.Activation('relu'),
    MaxPooling2D(2,2),

    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_bn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nTraining CNN with Batch Normalization")
model_bn.fit(x_train, y_train, epochs=5, batch_size=64)

loss, acc_bn = model_bn.evaluate(x_test, y_test)
print("Test Accuracy (Batch Normalization):", acc_bn)
```
